### check southarbica

```python
s_a_data = online_retail_data[online_retail_data["Country"]=="Saudi Arabia"]
s_a_data
```

```python
s_a_data
```

```python
s_a_pivotted_data= pd.pivot_table(data=s_a_data,values="Quantity",index="InvoiceNo",columns="Description",aggfunc="sum").fillna(0)
s_a_pivotted_data
```

```python
s_a_pivotted_onehott=s_a_pivotted_data.applymap(lambda x:0 if x==0 else 1)
s_a_pivotted_onehott
```

```python
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
frequent_itemsets_s_a = apriori(df = s_a_pivotted_onehott , min_support=0.02,use_colnames=True,)
frequent_itemsets_s_a
```

```python
best_associates_s_a = association_rules(df=frequent_itemsets_s_a,metric='confidence',min_threshold=0.02)
best_associates_s_a
```

```python
best_associates_s_a.to_csv("best_antecedent_consequent_s_a.csv")
```

### uk dataseta

```python
uk_pivotted_onehot= uk_pivotted_data.applymap(lambda x:0 if x==0 else 1)
uk_pivotted_onehot
```

### data mining

```python
from mlxtend.frequent_patterns import apriori,association_rules
```

```python
frequent_itemsets = apriori(df=uk_pivotted_onehot, min_support=0.02,use_colnames=True)
frequent_itemsets
```

```python
best_associates = association_rules(df=frequent_itemsets,metric='confidence',min_threshold=0.02)
best_associates
```

```python
best_associates.to_csv("best_antecedent_consequent.csv")
```
